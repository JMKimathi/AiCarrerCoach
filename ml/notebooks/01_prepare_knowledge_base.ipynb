# 01 — Prepare the career knowledge base

This notebook **does not train an LLM**. It turns career articles into chunks the retriever will search.

Run cells in order. Output: `ml/data/processed/chunks.json`

```python
from pathlib import Path
import json
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.append(str(ROOT / "src"))

from config import RESOURCES, CHUNKS, chunk_text

records = []
for path in sorted(RESOURCES.glob("*.md")):
    text = path.read_text(encoding="utf-8")
    title = text.strip().splitlines()[0].lstrip("# ").strip()
    resource_id = path.stem
    for i, chunk in enumerate(chunk_text(text)):
        records.append({
            "chunk_id": f"{resource_id}_{i}",
            "resource_id": resource_id,
            "title": title,
            "content": chunk,
        })

CHUNKS.parent.mkdir(parents=True, exist_ok=True)
CHUNKS.write_text(json.dumps(records, indent=2), encoding="utf-8")
print(f"Wrote {len(records)} chunks from {len(list(RESOURCES.glob('*.md')))} resources")
print("Saved to", CHUNKS)
records[:2]
```
